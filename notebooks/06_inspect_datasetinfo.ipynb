```python
from pathlib import Path
import numpy as np
```

```python
SEQ = Path.home() / \
"HDD/EventDatasets/EVIMO2_official/left_camera/imo/train/scene13_dyn_test_02_000000"

info = np.load(SEQ / "dataset_info.npz", allow_pickle=True)

print(type(info))
```

```python
print("="*80)
print("Keys")
print("="*80)

for key in info.files:
    print(key)

print()
print("Total keys:", len(info.files))
```

```python
for key in info.files:

    x = info[key]

    print("=" * 100)
    print(f"KEY : {key}")
    print("=" * 100)

    print("Type :", type(x))
    print("Shape:", x.shape)
    print("Dtype:", x.dtype)

    # ---------------------------------------------------------
    # Scalar objects (dict, list, etc.)
    # ---------------------------------------------------------
    if x.ndim == 0:

        value = x.item()

        print("\nScalar object")
        print("----------------")
        print("Python type:", type(value))

        if isinstance(value, dict):

            print("\nDictionary contains", len(value), "keys\n")

            for k in value.keys():
                print(" -", k)

        elif isinstance(value, (list, tuple)):

            print("\nLength:", len(value))

            for i, v in enumerate(value[:10]):
                print(i, type(v))

            if len(value) > 10:
                print("...")

        else:

            print(value)

    # ---------------------------------------------------------
    # Arrays
    # ---------------------------------------------------------
    else:

        print("\nArray statistics")
        print("----------------")
        print("Size :", x.size)

        if np.issubdtype(x.dtype, np.number):

            print("Min :", np.min(x))
            print("Max :", np.max(x))

        # ---------- 1D ----------
        if x.ndim == 1:

            print("\nFirst 10 values")
            print(x[:10])

            if len(x) > 10:
                print("Last 10 values")
                print(x[-10:])

        # ---------- 2D ----------
        elif x.ndim == 2:

            print("\nFirst 5 rows")
            print(x[:5])

        # ---------- Higher dimensions ----------
        else:

            print("\nFirst element")
            print(x[0])

    print()
```

```python
meta = info["meta"].item()

print("="*80)
print("META KEYS")
print("="*80)

for k in meta.keys():
    print(k)
```

```python
for key, value in meta.items():

    print("="*100)
    print(key)
    print("="*100)

    print("Type :", type(value))

    if isinstance(value, np.ndarray):
        print("Shape :", value.shape)
        print("Dtype :", value.dtype)

        if value.size < 20:
            print(value)
        else:
            print("First entries")
            print(value[:5])

    elif isinstance(value, dict):
        print("Dictionary with", len(value), "keys")

        for k in value.keys():
            print("-", k)

    else:
        print(value)

    print()
```
